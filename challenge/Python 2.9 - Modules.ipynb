```python
%%writefile data_utils.py
"""Fonctions utilitaires pour des calculs mathématiques et statistiques."""

import math

PI = math.pi
E = math.e


def carre(x):
    """Retourne le carré de x."""
    return x ** 2


def cube(x):
    """Retourne le cube de x."""
    return x ** 3


def factorielle(n):
    """Retourne la factorielle de l'entier n (n >= 0)."""
    if n < 0:
        raise ValueError("n doit être positif ou nul.")

    result = 1
    factor = n
    while factor > 1:
        result *= factor
        factor -= 1
    return result
        
        
def valeur_absolue(x):
    """Fonctions utilitaires pour des calculs mathématiques et statistiques."""
    if x < 0:
        return -x
        return x

def maximum(liste):
    """Retourne le plus grand élément de la liste."""
    maxi = liste[0]

    for item in liste[1:]:
        if item > maxi:
            maxi = item
    return maxi
        

def minimum(liste):
    """Retourne le plus petit élément de la liste."""
    if not liste:
        raise ValueError("La liste ne doit pas être vide.")

    smallest = liste[0]
    for item in liste[1:]:
        if item < smallest:
            smallest = item
    return smallest


def moyenne(liste):
    """Retourne la moyenne des éléments de la liste."""
    if not liste:
        raise ValueError("La liste ne doit pas être vide.")

    return sum(liste) / len(liste)


def ecart_type(liste):
    """Retourne l'écart type (population) des éléments de la liste."""
    if not liste:
        raise ValueError("La liste ne doit pas être vide.")

    average = moyenne(liste)
    total = 0
    for value in liste:
        total += (value - average) ** 2
    variance = total / len(liste)
    return math.sqrt(variance)

def mediane(liste):
    """Retourne la médiane des éléments de la liste."""
    if not liste:
        raise ValueError("La liste ne doit pas être vide.")
        
    sorted_list = sorted(liste)
    
    middle = len(sorted_list) // 2
    if len(sorted_list) % 2 == 1:
        return sorted_list
    return (sorted_list[middle - 1] + sorted_list[middle]) / 2    
```

## Etape 1 : Refaites le process qu'on a vu ensemble dans les quêtes et au début du challenge pour créer une nouvelle version du module `data_utils_V2.py` en rajoutant de nouvelles fonctions.


*   Création du fichier `data_utils_V2.py`
*   Ajout et développement de nouvelles fonctions :
  * `somme_cumulee(liste)`: calcule la somme cumulée d'une liste et retourne une liste. Par exemple, pour la liste [1, 2, 3, 4], la somme cumulée serait [1, 3, 6, 10].

  * `compter_occurrences(liste)`: retourne un dictionnaire contenant le nombre d'occurrences de chaque élément de la liste. Par exemple, pour la liste [1, 2, 2, 3], le résultat serait {1: 1, 2: 2, 3: 1}.
  
  * `inverser_dict(dictionnaire)`: inverse les clés et les valeurs d'un dictionnaire. Par exemple, pour le dictionnaire {'a': 1, 'b': 2}, le résultat serait {1: 'a', 2: 'b'}.

  * `filtrer_pairs(liste)`: retourne une nouvelle liste contenant uniquement les nombres pairs de la liste d'entrée. Par exemple, pour la liste [1, 2, 3, 4], le résultat serait [2, 4].

  * `celsius_to_fahrenheit(celsius)`: convertit une température de Celsius en Fahrenheit et retourne un nombre. Par exemple, pour une température de 0°C, le résultat serait 32°F. *Rappel : la formule est F = (C * 9/5) + 32*

  * `trouver_doublons(liste)`: retourne une liste des éléments qui apparaissent plus d'une fois dans la liste d'entrée. Par exemple, pour la liste [1, 2, 2, 3, 3], le résultat serait [2, 3].

  * `calculer_moyenne_ponderee(valeurs, poids)`: calcule la moyenne pondérée d'une liste de valeurs avec une liste de poids correspondante et retourne un nombre. Par exemple, pour les valeurs [3, 5] et les poids [1, 2], le résultat serait 4.33.





```python
%%writefile data_utils_V2.py
"""Version 2 des fonctions utilitaires : listes, dictionnaires et conversions."""

def somme_cumulee(liste):
    """Calcule la somme cumulée d'une liste et retourne une liste."""
    running_total = 0
    cumulative_sum = []
    for number in liste:
        running_total += number
        cumulative_sum.append(running_total)
    return cumulative_sum
        

def compter_occurrences(liste):
    """Retourne un dictionnaire contenant le nombre d'occurrences de chaque élément de la liste."""
    dictionnaire = {}
    for element in liste:
        if element in dictionnaire:
            dictionnaire[element] += 1
        else:
            dictionnaire[element] = 1
    return dictionnaire
        

def inverser_dict(dictionnaire):
    """Inverse les clés et les valeurs d'un dictionnaire."""
    dict_inverse = {}
    for clé, valeur in dictionnaire.items():
        dict_inverse[valeur] = clé
    return dict_inverse


def filtrer_pairs(liste):
    """Retourne une nouvelle liste contenant uniquement les nombres pairs."""
    nouvelle_liste = []
    for valeur in liste:
        if valeur % 2 == 0:
            nouvelle_liste.append(valeur)
    return nouvelle_liste

def celsius_to_fahrenheit(celcius):
    """Convertit une température de Celcius en Fahrenheit et retourne un nombre."""
    F = (celcius * 9/5) + 32
    return F
    

def trouver_doublons(liste):
    """Retourne une liste des éléments qui apparaissent plus d'une fois dans la liste d'entrée."""
    liste_doublon = []
    elements = []
    for valeur in liste:
        if valeur in elements and valeur not in liste_doublon:
            liste_doublon.append(valeur)
        else: 
            elements.append(valeur)
    return liste_doublon
            

def calculer_moyenne_ponderee(valeurs, poids):
    """Calcule la moyenne pondérée d'une liste de valeurs avec 
    une liste de poids correspondante et retourne un nombre."""
    produit_listes = 0
    somme_poids = sum(poids)
                     
    if len(valeurs) != len(poids):
        raise ValueError("Les deux listes doivent avoir la même taille.")
    if somme_poids == 0:
        raise ValueError("La somme des poids ne doit pas être nulle.")

    for value, weight in zip(valeurs, poids):
            produit_listes += value * weight
        
    return produit_listes / somme_poids    
```

# Partie Bonus

Créez une fonction `analyser_texte(texte)` qui prend une chaîne de caractères en entrée et retourne un dictionnaire contenant les informations suivantes :

- Nombre total de mots
- Nombre de caractères (espaces inclus)
- Nombre de caractères (espaces exclus)
- Mot le plus long
- Longueur moyenne des mots

Testez cette fonction avec le texte suivant :
"Python est un langage de programmation puissant et facile à apprendre. Il dispose de structures de données de haut niveau efficaces et d'une approche simple mais efficace de la programmation orientée objet."

```python
texte_test = "Python est un langage de programmation puissant et facile à apprendre. Il dispose de structures de données de haut niveau efficaces et d'une approche simple mais efficace de la programmation orientée objet."
```

```python
def analyser_texte(texte):
    mots = texte.strip(".").split()
    nombre_mots = len(mots)
    nombre_caracteres = len(texte)
    nombre_caracteres_sans_espaces = len(list(texte)) - texte.count(" ")
    mot_le_plus_long = max(mots, key=len)
    longueur_moyenne_mots = round((sum([len(mot) for mot in mots])) / nombre_mots)
    return nombre_mots, nombre_caracteres, nombre_caracteres_sans_espaces, mot_le_plus_long, longueur_moyenne_mots
```
