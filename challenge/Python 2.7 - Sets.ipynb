Question : Why isn't `my_set` an object from the `dict` class ?

Answer : Both use curly braces, but a dict is made of key: value pairs ({"a": 1}), and my_set only has single values. Python decides by the content. An empty {} is a dict, so an empty set must be written set().

Question :  

Can a set contain elements with differents types ?

Yes. Your example {"hey", 2, True, 2.33} works. Two caveats:

True equals 1 and 2.0 equals 2 for a set, so {1, True, 1.0} keeps one element: {1}.
Elements must be hashable (immutable). A tuple is allowed, but a list or a dictionary inside a set raises a TypeError.

```python
# try to make a set with objects with different types here:
answer = {"hey", 2, True, 2.0}
print(answer)
```

There are four collection data types in the Python programming language:

* `List` is a collection which is ordered and changeable. Allows duplicate members.
* `Tuple` is a collection which is ordered and unchangeable. Allows duplicate members.
* `Set` is a collection which is unordered, unchangeable, and unindexed. No duplicate members, but you can remove items and add new items.
* `Dictionary` is a collection which is ordered and changeable. No duplicate members.

Question :    

If you display a `set` object, it won't display with the same order you define it.

What can you say about the order the element are display ?

Answer : A set is **unordered**: Python stores elements by a hash value, not by the order in which you wrote them. The display order can differ from your definition, and for strings it can even change between two runs of Python. Never rely on it. If you need an order, use sorted(my_set), which gives a sorted list.
