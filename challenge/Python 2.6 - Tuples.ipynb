# Mission 4

Create 2 tuples of the same length that contain only integers.

Then create a script that compares the sum of their elements and displays the tuple that has the highest total.

```python
tuple_1 = (4, 5, 6)
tuple_2 = (-10, 15, 20)
```

```python
def compare_tuple_sums(first_tuple, second_tuple):
    """Return which tuple has the highest sum."""
    first_sum = sum(first_tuple)
    second_sum = sum(second_tuple)

    if first_sum == second_sum:
        return f"The tuples have the same sum: {first_sum}"
    if first_sum > second_sum:
        return f"The first tuple has the highest total: {first_sum}"
    return f"The second tuple has the highest total: {second_sum}"
```
