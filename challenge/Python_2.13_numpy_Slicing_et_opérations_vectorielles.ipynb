## NumPy cheat sheet

### Basics

| Goal | Code | Notes |
|---|---|---|
| Import | `import numpy as np` | Always `np`, at the top |
| Version | `np.__version__` | Attribute, no parentheses |

### Creating arrays

| Goal | Code | Result or note |
|---|---|---|
| From a list | `np.array([0.5, 1.5])` | One single type for all elements |
| Only zeros | `np.zeros(4)` | `[0., 0., 0., 0.]` |
| Only ones | `np.ones(3)` | `[1., 1., 1.]` |
| Reserved but not filled | `np.empty(3)` | Garbage values from memory, faster than `zeros` |
| Range of numbers | `np.arange(0, 10, 2)` | `[0, 2, 4, 6, 8]`, end excluded |
| 2D array | `np.zeros((2, 3))` | The size is a tuple |

### Characteristics

| Goal | Code | Example result |
|---|---|---|
| Type of elements | `a.dtype` | `float64` |
| Size of each axis, as a tuple | `a.shape` | `(3,)` |
| Number of axes | `a.ndim` | `1` |
| Number of elements | `len(a)` | `3` for a flat array |
| Total number of elements | `a.size` | `6` for shape `(2, 3)` |

### Shape and axes

| Goal | Code | Notes |
|---|---|---|
| Change the shape | `a.reshape(rows, cols)` | Rows first, then columns |
| Same, as a function | `np.reshape(a, (rows, cols))` | The new shape is a tuple here |
| Let NumPy find one size | `a.reshape(2, -1)` | `-1` means "calculate it" |
| 1 row, n columns | `a.reshape(1, n)` | Shape `(1, n)`, double brackets `[[ ]]` |
| n rows, 1 column | `a.reshape(n, 1)` | Shape `(n, 1)` |

Rules for `reshape`:

- The product of the new sizes must equal the number of elements, otherwise you get a `ValueError`.
- It **returns a new array** and does not change the original. Save the result in a variable.
- Shape `(32,)` has one axis. Shape `(1, 32)` has two axes.

### Indexing and filtering

| Goal | Code | Example result |
|---|---|---|
| First element | `a[0]` | `10` |
| Last element | `a[-1]` | `50` |
| Slice | `a[1:4]` | Indexes 1 to 3, end excluded |
| Filter with a condition | `a[a > 25]` | Only the elements above 25 |
| Element of a 2D array | `b[row, col]` | `b[0, 2]` is row 0, column 2 |

### Random numbers

| Goal | Code | Notes |
|---|---|---|
| Floats between 0 and 1 | `np.random.random(n)` | 0 included, 1 excluded |
| Random integers | `np.random.randint(low, high, size)` | `high` excluded, integers only |
| Pick among values | `np.random.choice(values, size)` | Repeats allowed by default |
| Floats between `a` and `b` | `a + (b - a) * np.random.random(n)` | Scale, then shift |

### Operations on the whole array (vectorization)

| Goal | Code | Notes |
|---|---|---|
| Math on every element | `a * 2`, `a + 1`, `a ** 2` | No loop needed |
| Sum | `a.sum()` | |
| Mean | `a.mean()` | |
| Minimum and maximum | `a.min()`, `a.max()` | |
| Standard deviation | `a.std()` | Population version by default |

### Reminders

| Point | Reminder |
|---|---|
| List versus array | `[1, 2] * 2` repeats the list, `np.array([1, 2]) * 2` multiplies each element |
| One type only | A single float turns the whole array into floats |
| End excluded | `arange`, slices and `randint` all exclude the end |
| Methods that return a new array | `reshape` does not modify the original, so assign the result |
| Conventions | Clear names, never `np`, `array` or `random` as variable names, one-line docstring, `return` and not `print` in functions |

Rules for `reshape`:

Modify, using a `numpy` method, the `array_1` object to be an object that has two axes, including 1 row and 32 columns.  
Save it under the variable `array_2`.

```python
array_2 = array_1.reshape(1, 32)
print(array_2)
```

Now, modify `array_1` into an object that has 32 rows and only 1 column.  
Save it under the variable `array_3`.

```python
array_3 = array_1.reshape(32, 1)
print(array_3)
```

### Indexing and filtering

| Goal | Code | Example result |
|---|---|---|
| First element | `a[0]` | `10` |
| Last element | `a[-1]` | `50` |
| Slice | `a[1:4]` | Indexes 1 to 3, end excluded |
| Filter with a condition | `a[a > 25]` | Only the elements above 25 |
| Element of a 2D array | `b[row, col]` | `b[0, 2]` is row 0, column 2 |

Access the first element of `array_2`, then of `array_3`.  
What do you notice? 
==> With a 2D array, a[0] does not give the first element. It gives the first row.

Why ?
==> The index in a[0] is applied to the first axis, which is the rows axis. In array_2, there is only one row, and it contains everything. In array_3, there are 32 rows of one value each, so the first row is just [10].

In both cases, the result is still an array, not a plain number. This is shown by the brackets: array([10]) and not 10. A 1D array like array_1 has no row axis, so array_1[0] gives the number 10 directly.

```python
array_2[0]
```

```python
array_3[0]
```
