# 1. Calcul d'un prix

Ecrivez un programme faisant saisir à l’utilisateur un prix unitaire HT, un taux de TVA et un nombre d’articles, et qui calcule le montant TTC de l’achat.<br>
Affichez le montant TTC.

```python
def montantTTC():
    prix = float(input())
    tva = float(input())
    article = int(input())
    return prix * (1 + tva) * article
```

```python
montantTTC()
```

```python
# CONVENTION
def compute_total_with_vat(unit_price, vat_rate, article_count):
    """Return the total price including VAT."""
    return unit_price * (1 + vat_rate) * article_count
```

```python
price = float(input("Prix unitaire HT : "))
rate = float(input("Taux de TVA (par exemple 0.2 pour 20 %) : "))
quantity = int(input("Nombre d'articles : "))

total = compute_total_with_vat(price, rate, quantity)
print(f"Montant TTC : {total:.2f}")
```

# 2. Conversion de durées (1)

Ecrivez un programme demandant une durée en secondes à l’utilisateur, et qui la convertit en heures, minutes, secondes.<br>
Affichez le résultat.<br>
Par exemple 12334 deviendra 3 heures, 25 minutes et 34 secondes.

```python
def convertduree():
    dureesec = int(input("Ecris une durée en seconde et je te la renvoie en heures, minutes et secondes"))
    
    heure = int(dureesec/60/60)
    minute = int((dureesec - (heure*60*60))/60)
    seconde = int(dureesec - (heure*60*60) - (minute*60))
    
    print(f"{dureesec} devient {heure} heure(s), {minute} minute(s) et {seconde} seconde()s.")
```

```python
convertduree()
```

```python
# CONVENTION
SECONDS_PER_MINUTE = 60
SECONDS_PER_HOUR = 3600

def convert_seconds_to_hms(total_seconds):
    """Return a duration in seconds as (hours, minutes, seconds)."""
    hours = total_seconds // SECONDS_PER_HOUR
    remaining_seconds = total_seconds % SECONDS_PER_HOUR
    minutes = remaining_seconds // SECONDS_PER_MINUTE
    seconds = remaining_seconds % SECONDS_PER_MINUTE
    return hours, minutes, seconds
```

```python
duration = int(input("Durée en secondes : "))

hours, minutes, seconds = convert_seconds_to_hms(duration)
print(f"{hours} heures, {minutes} minutes et {seconds} secondes")
```

# 3. Conversion de durées (2)

Ecrivez un programme demandant une durée en heures, minutes, secondes à l’utilisateur et qui la convertit en secondes.<br>
Affichez le résultat.<br>
Par exemple 3 heures, 25 minutes et 34 secondes deviendra 12334 secondes.

```python
def convertduree2():
    print("Rentre une durée en heures, minutes et secondes et je la convertis uniquement en seconde.")
    heures = int(input("heure(s)"))
    minutes = int(input("minute(s)"))
    secondes = int(input("seconde(s)"))

    dureesec = int(heures*60*60 + minutes*60 + secondes)
    return dureesec
    
```

```python
convertduree2()
```

```python
# CONVENTION

SECONDS_PER_MINUTE = 60
SECONDS_PER_HOUR = 3600

def convert_hms_to_seconds(hours, minutes, seconds):
    """Return a duration given as hours, minutes and seconds, in seconds."""
    return hours * SECONDS_PER_HOUR + minutes * SECONDS_PER_MINUTE + seconds
```

```python
hours = int(input("Heures : "))
minutes = int(input("Minutes : "))
seconds = int(input("Secondes : "))

duration = convert_hms_to_seconds(hours, minutes, seconds)
print(f"{duration} secondes")
```
