## NumPy cheat sheet

### Basics
| Goal | Code | Notes |
|---|---|---|
| Import | `import numpy as np` | Always `np`, at the top |
| Version | `np.__version__` | Attribute, no parentheses |

### Creating arrays

| Goal | Code | Result or note |
|---|---|---|
| From a list | `np.array([0.5, 1.5])` | One single type for all elements |
| Only zeros | `np.zeros(4)` | `[0., 0., 0., 0.]` |
| Only ones | `np.ones(3)` | `[1., 1., 1.]` |
| Reserved but not filled | `np.empty(3)` | Garbage values from memory, faster than `zeros` |
| Range of numbers | `np.arange(0, 10, 2)` | `[0, 2, 4, 6, 8]`, end excluded |
| 2D array | `np.zeros((2, 3))` | The size is a tuple |

### Characteristics

| Goal | Code | Example result |
|---|---|---|
| Type of elements | `a.dtype` | `float64` |
| Size as a tuple | `a.shape` | `(3,)` |
| Number of dimensions | `a.ndim` | `1` |
| Number of elements | `len(a)` | `3` |

### Indexing and filtering

| Goal | Code | Example result |
|---|---|---|
| First element | `a[0]` | `10` |
| Last element | `a[-1]` | `50` |
| Slice | `a[1:4]` | Indexes 1 to 3, end excluded |
| Filter with a condition | `a[a > 25]` | Only the elements above 25 |

### Random numbers

| Goal | Code | Notes |
|---|---|---|
| Floats between 0 and 1 | `np.random.random(n)` | 0 included, 1 excluded |
| Random integers | `np.random.randint(low, high, size)` | `high` excluded |
| Pick among values | `np.random.choice(values, size)` | Repeats allowed by default |
| Floats between `a` and `b` | `a + (b - a) * np.random.random(n)` | Scale, then shift |

### Operations on the whole array (vectorization)

| Goal | Code | Notes |
|---|---|---|
| Math on every element | `a * 2`, `a + 1`, `a ** 2` | No loop needed |
| Sum | `a.sum()` | |
| Mean | `a.mean()` | |
| Minimum and maximum | `a.min()`, `a.max()` | |
| Standard deviation | `a.std()` | Population version by default |

### Reminders

| Point | Reminder |
|---|---|
| List versus array | `[1, 2] * 2` repeats the list, `np.array([1, 2]) * 2` multiplies each element |
| One type only | A single float turns the whole array into floats |
| End excluded | `arange`, slices and `randint` all exclude the end |
| Conventions | Clear names, never `np`, `array` or `random` as variable names, one-line docstring, `return` and not `print` in functions |
