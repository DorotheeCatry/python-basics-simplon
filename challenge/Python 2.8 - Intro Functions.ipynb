# Mission 2

Create a function that returns the sum of two numbers.  
The user of the function must be able to sum two numbers of his choice, when the function is executed.  
You must not use `input`.  

```python
def sum(first_num, second_num):
    return first_num + second_num
```

```python
sum(3, 3)
```

Question:   
What is a parameter in a function?

A parameter is a variable name written in the parentheses of a fonction definition. It is a placeholder for a value that the function will receive when someone calls it.

# Mission 3

Create a function that returns the sum of two numbers chosen by the user of the function, in a sentence.  
For example:  
`new_sum_2(3,8)` returns `"The sum of 3 + 8 is equal to 11."`

```python
def sum_sentence(first_num, second_num):
    return f"The sum of {first_num} + {second_num} is equal to {first_num + second_num}"
```

```python
sum_sentence(3, 8)
```

# Mission 4

Create a function that returns 2 objects when executed:
* An object that will be the product of two numbers chosen by the user of the function.
* A word chosen by the user of the function.

```python
def product_of(first_num, second_num, word):
    first_object = first_num * second_num
    return first_object, word
```

```python
product_of(3, 8, "mot")
```

Your result must be two objects that are in brackets.  
This is a tuple.

Question:  
Why is the returned object a tuple that contains two objects?  

Answer: A function can only return one object. When we write return with two values separated by a comma, Python packs them together in a tuple, which is one object containing the two values.


# Mission 5

Create a function that returns the sum of two numbers when executed.  
Also, your function must return a sentence that simply indicates if the sum of the two numbers chosen by the user is:
* less than 10
* between 10 and 50
* greater than 50  
Again, you must use `return`.

```python
def sum_with_description(first_num, second_num):
    sum_num = first_num + second_num
    if sum_num < 10:
        return f"{sum_num}, the sum is less than 10"
    elif sum_num <= 50:
        return f"{sum_num}, the sum is between 10 and 50"
    else: 
        return f"{sum_num}, the sum is greater than 50"
```

```python
sum_with_description(5, 2)
```
