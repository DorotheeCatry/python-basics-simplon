What's the lenght of the following string ?

````python
s = "Taumata whakatangihanga koauau o tamatea turi pukaka piki maungah oronuku pokai whenuaki tanatahu "
len(s)
````

Now, count only non-space characters and display how many there is.
Remember, using a for loop on a string will browse all the individual characters of it!

````python
s = "Taumata whakatangihanga koauau o tamatea turi pukaka piki maungah oronuku pokai whenuaki tanatahu "

def non_space_count(text):
    n = 0
    for character in text:
        if character != " ":
            n += 1
    return n
        
````

````python
non_space_count(s)
````

# Mission 2

Ask the user for a string, then for a character (we assume user input follows the rules). Then, display the positions, **starting with 1**, of all occurrences of the character within the string.

For example, for the word "welcome" and for the character "e", the script will display:
```
position: 2
position: 7
```

````python
def find_the_string(text, character):
    n = 0
    for char in text:
        n += 1
        if char == character:
            print(f"position: {n}")
````

````python
find_the_string("welcome", "e")
````

````python
def print_positions(text, character):
    """Print the positions (starting at 1) of a character in a text."""
    position = 0
    for current_character in text:
        position += 1
        if current_character == character:
            print(f"position: {position}")
````

````python
user_text = input("Enter a string: ")
user_character = input("Enter a character: ")
print_positions(user_text, user_character)
````

# Mission 3

Ask the user for a string, then for an integer. Then, display the string repeated as many times as the given integer.

For example, `badger` and `4` will display `badgerbadgerbadgerbadger`

````python
def repeat(text, times):
    return text * int(times)
````

````python
user_text = input("Enter a string: ")
user_character = input("Enter an integer: ")
repeat(user_text, user_character)
````

# Mission 5

Ask the user for a string, then for a character, then for another character.

Then, display the string where the occurences of the first character are replaced by the second one.  
For example:
* string = `"Boulgour"`  
* 1st_char = `"g"`
* 2nd_char = `"v"`

Result : `"Boulvour"`

````python
def char_replace(text, char_1, char_2):
    return text.replace(char_1, char_2)
````

````python
user_text = input("Enter a string: ")
char_1 = input("Enter an integer: ")
char_2 = input("Enter an integer: ")

char_replace(user_text, char_1, char_2)
````

# Mission 6

Ask the user for a string. Then, display the same string with all vowels converted to uppercase, and all consonants converted to lowercase.  
For example:
* string = `"Papa's got a brand new bag"`  
Result : `"pApA's gOt A brAnd nEw bAg"`


````python
def char_up_down(text):
    result = ""
    for char in text:
        if char in "aiueoAIUEO":
            result += char.upper()
        else:
            result += char.lower()
    return result
````

````python
user_text = input("Enter a string: ")

char_up_down(user_text)
````

````python
# CONVENTION
VOWELS = "aeiou"


def upper_vowels_lower_consonants(text):
    """Return a text with vowels in uppercase and other letters in lowercase."""
    result = ""
    for character in text:
        if character.lower() in VOWELS:
            result += character.upper()
        else:
            result += character.lower()
    return result
````

````python
user_text = input("Enter a string: ")
print(upper_vowels_lower_consonants(user_text))
````
