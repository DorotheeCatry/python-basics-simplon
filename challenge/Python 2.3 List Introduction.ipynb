For each cell, execute the code and answer the question before moving on to the next.

```python
names = ["Fredy", "Nicolas", "Alexandre"]

for i in enumerate(names):
  print(i)
```

1. What type of object is returned as output?  
Hint: You can use `type(i)`.

Answer: A tuple. Each i is a pair (index, element).


```python
for i, val in enumerate(names):
  print(val)
```

2. What does `i` correspond to?  
What does `val` correspond to?  
Why is it so?

Answer: i is the index (the position in the list) and val is the value (the element at that position). It is so because enumerate gives pairs (index, element), and Python assigns the first item of the pair to i and the second one to val.

```python
for val, i in enumerate(names):
  print(val)
```

3. This time, `i` returns first names and `val` returns numbers. Why is it so?

Answer : Because Python unpacks by order, not by name. The first item of each pair is always the index, so the first variable receives the index, whatever its name. Here the first variable is called val, so val gets the numbers. The names are only labels, so they should describe what the variables really contain.

4. What does the built in function `enumerate` do?
In what cases can it be useful?

Answer: enumerate adds a counter to an iterable and returns pairs (index, element). It is useful when we need both the position and the value in a loop, for example to number items in a list, to find the positions of an element, or to avoid managing a manual counter. With enumerate(names, start=1), the counter starts at 1.

```python
test_list = ['Hi', 'my', 'name', 'is', "hi", "my","name", "is", "chickychicky"]

test_list[8] = "Marshall"

test_list
```

1. What does the 8 in square brackets mean?  
What action has been performed?

Answer: It is the index of an element in the list. Indexes start at 0, so index 8 is the ninth and last element. The action is a replacement: "chickychicky" was replaced by "Marshall". It works because lists are mutable.

```python
for i in range(4):
  print(test_list[i])
```

2. What does `range(4)` mean?  
Why are only the first 4 elements of `test_list` displayed?

Answer: range(4) generates the integers 0, 1, 2 and 3 (it starts at 0 and the end, 4, is excluded). The loop uses them as indexes, so it displays the elements at positions 0 to 3, which are the first four.

```python
for i in range(2, 8, 2):
  print(test_list[i])
```

3. What does `range(2,8,2)` mean ?  
Why are only the elements 'name', 'hi' and 'name' displayed?

Answer : it means: start at 2, stop before 8, step of 2. It generates 2, 4 and 6. These are the indexes used, and test_list[2], test_list[4] and test_list[6] are 'name', 'hi' and 'name'.

1. What can the `.join()` method do?

What is the type of `test_list` after using `.join()`?

2. What can the `.split()` method do?  
What is the type of `christmas` before using `.split()`?

Answer: .split() cuts a string into pieces and returns them in a list. By default, it cuts at each space. christmas is a str before. 3. Why does the `.split()` method split the `string` into 3 elements?
