# 1. Calcul du montant d'une remise

Un commerçant accorde une remise de 5 % pour tout achat d’un montant compris entre 100 et 500 € et 8 % au-delà.<br>
Ecrivez un programme de calcul du montant de la remise sur un achat donné.<br>
Affichez le montant de la remise.

```python
def remise():
    prixachat = int(input("Achat :"))
    if prixachat < 100:
        return print("Pas de remise")
    elif 100 <= prixachat <= 500:
        print("Remise de 5%")
        return prixachat*0.05
    else:
        print("Remise de 8%")
        return prixachat*0.02
        
```

```python
remise()
```

```python
# CONVENTION 

LOW_THRESHOLD = 100
HIGH_THRESHOLD = 500
LOW_RATE = 0.05
HIGH_RATE = 0.08


def compute_discount(purchase_amount):
    """Return the discount amount for a purchase."""
    if purchase_amount < LOW_THRESHOLD:
        return 0
    if purchase_amount <= HIGH_THRESHOLD:
        return purchase_amount * LOW_RATE
    return purchase_amount * HIGH_RATE
```

# 2. Savoir si trois entiers sont triés

Ecrivez un programme faisant saisir trois entiers x, y, z à l’utilisateur, et lui indiquer si ces nombres sont dans l’ordre croissant (x <= y <= z).

```python
def tri():
    print("Saisissez 3 entiers pour x, y et z et je te dirai s'ils sont dans l'ordre croissant")
    x = int(input("x = "))
    y = int(input("y = "))
    z = int(input("z = "))

    if  x <= y <= z :
        return print("Les chiffres sont dans un ordre croissant.")
    else:
        return print("Les chiffres ne sont pas dans un ordre croissant.")
    
```

```python
tri()
```

```python
# CONVENTION

def is_sorted(first, second, third):
    """Return True if the three numbers are in ascending order."""
    return first <= second <= third
```

```python
x = int(input("x = "))
y = int(input("y = "))
z = int(input("z = "))

if is_sorted(x, y, z):
    print("Les nombres sont dans l'ordre croissant.")
else:
    print("Les nombres ne sont pas dans l'ordre croissant.")
```

# 3. Tri de trois réels

Ecrivez un programme faisant saisir trois nombres réels x, y, z à l’utilisateur et qui les trie par ordre croissant (à la fin du déroulement du programme x ≤ y ≤ z).<br>
Affichez x, y et z.

```python
def tri2():
    print("Saisissez 3 entiers pour x, y et z et je te les tri par ordre croissant.")
    x = float(input("x = "))
    y = float(input("y = "))
    z = float(input("z = "))

    if x > y:
        (x, y) = (y, x)
    if x > z:
        (x, z) = (z, x)
    if y > z:
        (y, z) = (z, y)
    return print(f"x(={x}) <= y(={y}) <= z(={z})")
```

```python
tri2()
```

```python
# CONVENTION

def sort_three(first, second, third):
    """Return three numbers sorted in ascending order."""
    if first > second:
        first, second = second, first
    if first > third:
        first, third = third, first
    if second > third:
        second, third = third, second
    return first, second, third
```

```python
x = float(input("x = "))
y = float(input("y = "))
z = float(input("z = "))

x, y, z = sort_three(x, y, z)
print(f"{x} <= {y} <= {z}")
```

# 4. Signe d'un produit

Ecrivez un programme qui affiche le signe du produit de deux nombres réels sans calculer la valeur de ce produit. Par signe, on entend positif, négatif ou nul.

```python
def signe():
    x = float(input("x = "))
    y = float(input("y = "))

    if x == 0 or y == 0:
        print("null")
    elif (x < 0 and y < 0) or (x > 0 and y > 0):
        print("+")
    else:
        print("-")
    
```

```python
signe()
```

```python
# CONVENTION

def get_product_sign(first, second):
    """Return the sign of the product of two numbers, without computing it."""
    if first == 0 or second == 0:
        return "nul"
    if (first > 0 and second > 0) or (first < 0 and second < 0):
        return "positif"
    return "négatif"
```

```python
x = float(input("x = "))
y = float(input("y = "))

print(get_product_sign(x, y))
```
